## **4. Matriz de correlación**

## **5. Función de conversión**

Crea una función para convertir los valores de la temperatura del dataset que se encuentran en grados Celsius a grados Fahrenheit y agrega la conversión en una nueva columna.

```python
# Definimos la función que convierte de Celsius a Fahrenhei
def celsius_a_fahrenheit(c):
    return (c * 9/5) + 32
```

```python
# Definimos una nueva columna donde se guardará la conversión
clima_ds['Temperature_Fahrenheit'] = celsius_a_fahrenheit(clima_ds['Temperature'])
```

## **7. Diagrama de dispersión**

## **8. Histograma.**

Usando Matplotlib, representa el histograma para las características precipitación y humedad usando 30 divisiones con color azul. Muestra la media y la desviación estándar para ambas acciones en la parte superior de la figura.

```python

# Definimos el tamaño del gráfico
plt.figure(figsize=(12, 5))

# Calculamos la media y desviación estándar de precipitación
media_precipitacion = clima_ds['Precipitation (%)'].mean()
desv_precipitacion = clima_ds['Precipitation (%)'].std()

# Calculamos la media y desviación estándar de humedad
media_humedad = clima_ds['Humidity'].mean()
desv_humedad = clima_ds['Humidity'].std()

# Graficamos los histogramas
plt.hist(clima_ds['Precipitation (%)'], bins=30, color='blue', alpha=0.5,
         label=f'Precipitación: media={media_precipitacion:.2f}, desviación={desv_precipitacion:.2f}')

plt.hist(clima_ds['Humidity'], bins=30, color='blue', alpha=0.5,
         label=f'Humedad: media={media_humedad:.2f}, desviación={desv_humedad:.2f}')

plt.title('Histograma de Precipitación y Humedad')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.legend()
plt.show()
```

 Representa un gráfico 3D que muestre todos los valores de la temperatura, humedad y precipitación.

## **10. Escalamiento y normalización.**

Haz el escalamiento de las características temperatura, humedad, velocidad del viento y precipitación. Escala las características presión atmosférica, UV index y visibilidad utilizando la normalización.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
clima_scaled = clima_ds.copy()
```

```python
# Cargamos la librería preprocessing para el escalamiento de datos
scaler = StandardScaler()
cols_to_scale = ['Temperature', 'Humidity', 'Wind Speed', 'Precipitation (%)']
clima_scaled[cols_to_scale] = scaler.fit_transform(clima_scaled[cols_to_scale])
```

```python
# Cargamos la librería preprocessing para el escalamiento de datos
normalizer = MinMaxScaler()
cols_to_norm = ['Atmospheric Pressure', 'UV Index', 'Visibility (km)']
clima_scaled[cols_to_norm] = normalizer.fit_transform(clima_scaled[cols_to_norm])
```

```python
# Mostramos los valores estadísticos para verificar los valores
display(clima_scaled[['Temperature', 'Atmospheric Pressure']].describe())
```

## **12. Gráficas de burbuja.**

Crea un gráfico de burbuja definiendo una característica para el uso de colores.

## **14. Gráfico de pila.**

Crea un gráfico stacked para representar datos del dataset.

```python
# Agrupamos los datos por Cloud cover y weather type.
pivot_data = pd.crosstab(clima_ds['Cloud Cover'], clima_ds['Weather Type'])
```

```python
# Gráfico de Barras apiladas de temperatura por tipo de clima y tipo de nubes
pivot_data.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='tab10')
plt.title("Gráfico de Pila: Weather Type según Cloud Cover")
plt.xlabel("Cloud Cover")
plt.ylabel("Frecuencia")
plt.xticks(rotation=0)
plt.legend(title='Weather Type', loc='upper left')
plt.show()
```
